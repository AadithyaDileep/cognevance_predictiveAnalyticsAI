```python
# USER MOVIE MATRIX CREATION

user_movie_matrix = ratings.pivot_table(
    index="user_id",
    columns="movie_id",
    values="rating"
)

print(
    "User-Movie Matrix Shape:",
    user_movie_matrix.shape
)

display(user_movie_matrix.head())
```

```python
# FILLING MISSING VALUES

user_movie_matrix_filled = user_movie_matrix.fillna(0)

display(
    user_movie_matrix_filled.head()
)
```

```python
# MOVIE SIMILARITY CALCULATION

movie_similarity = cosine_similarity(
    user_movie_matrix_filled.T
)

print(
    "Movie similarity matrix shape:",
    movie_similarity.shape
)
```

```python
movie_similarity_df = pd.DataFrame(
    movie_similarity,
    index=user_movie_matrix.columns,
    columns=user_movie_matrix.columns
)

movie_similarity_df.head()
```

```python
# MOVIE LOOKUP DICTIONARY CREATION

movie_id_to_title = dict(
    zip(
        movies["movie_id"],
        movies["title"]
    )
)

title_to_movie_id = dict(
    zip(
        movies["title"],
        movies["movie_id"]
    )
)

print("Movie lookup created.")
```

```python
# RECOMMENDATION FUNCTION CREATION

def recommend_movies(movie_title, n=10):
    """
    Recommend movies similar to the given movie
    using item-based collaborative filtering.
    """

    # Find matching movie
    matches = movies[
        movies["title"].str.contains(
            movie_title,
            case=False,
            na=False
        )
    ]

    if matches.empty:
        print("Movie not found.")
        return None

    # Select first matching movie
    selected_movie = matches.iloc[0]

    movie_id = int(selected_movie["movie_id"])

    print(
        f"Selected movie: {selected_movie['title']}"
    )

    # Get similarity scores
    similarity_scores = movie_similarity_df[movie_id]

    # Sort similarity scores
    similar_movies = similarity_scores.sort_values(
        ascending=False
    )

    # Remove the selected movie
    similar_movies = similar_movies[
        similar_movies.index != movie_id
    ]

    # Select top N
    top_movies = similar_movies.head(n)

    # Create result
    result = pd.DataFrame({
        "movie_id": top_movies.index,
        "similarity_score": top_movies.values
    })

    # Add movie titles
    result["title"] = result["movie_id"].map(
        movie_id_to_title
    )

    result = result[
        [
            "movie_id",
            "title",
            "similarity_score"
        ]
    ]

    return result.reset_index(drop=True)
```

```python
# USER RECOMMENDATION SYSTEM CREATION

user_similarity = cosine_similarity(
    user_movie_matrix_filled
)

user_similarity_df = pd.DataFrame(
    user_similarity,
    index=user_movie_matrix_filled.index,
    columns=user_movie_matrix_filled.index
)

print(
    "User similarity matrix:",
    user_similarity_df.shape
)
```

```python
# USER RECOMMENDATION FUNCTION

def recommend_for_user(user_id, n=10):

    if user_id not in user_movie_matrix.index:
        print("User ID not found.")
        return None

    # Find similar users
    similar_users = (
        user_similarity_df[user_id]
        .sort_values(ascending=False)
    )

    # Remove current user
    similar_users = similar_users[
        similar_users.index != user_id
    ]

    # Select top similar users
    top_users = similar_users.head(10)

    # Get ratings from similar users
    candidate_ratings = ratings[
        ratings["user_id"].isin(top_users.index)
    ].copy()

    # Movies already watched by current user
    watched_movies = ratings[
        ratings["user_id"] == user_id
    ]["movie_id"].tolist()

    # Remove watched movies
    candidate_ratings = candidate_ratings[
        ~candidate_ratings["movie_id"].isin(
            watched_movies
        )
    ]

    if candidate_ratings.empty:
        print("No recommendations available.")
        return None

    # Calculate predicted rating
    recommendations = (
        candidate_ratings
        .groupby("movie_id")["rating"]
        .mean()
        .reset_index()
    )

    recommendations.columns = [
        "movie_id",
        "predicted_rating"
    ]

    # Sort
    recommendations = recommendations.sort_values(
        "predicted_rating",
        ascending=False
    ).head(n)

    # Add movie titles
    recommendations = recommendations.merge(
        movies[["movie_id", "title"]],
        on="movie_id",
        how="left"
    )

    return recommendations[
        [
            "movie_id",
            "title",
            "predicted_rating"
        ]
    ].reset_index(drop=True)
```

```python
# POPULATIRY BASED FALLBACK CREATION

def get_popular_movies(n=10, minimum_ratings=50):

    movie_data = (
        ratings
        .groupby("movie_id")
        .agg(
            average_rating=("rating", "mean"),
            rating_count=("rating", "count")
        )
        .reset_index()
    )

    # Minimum number of ratings
    movie_data = movie_data[
        movie_data["rating_count"] >= minimum_ratings
    ]

    # Add titles
    movie_data = movie_data.merge(
        movies[["movie_id", "title"]],
        on="movie_id",
        how="left"
    )

    # Sort
    movie_data = movie_data.sort_values(
        ["average_rating", "rating_count"],
        ascending=[False, False]
    )

    return movie_data.head(n)
```
